### Reshape array

```python
x = np.array([1, 2, 3, 4, 5, 6])
print(f'beginning x:\n {x}')
x_reshape = x.reshape((2, 3))
print(f'reshaped x:\n {x_reshape}')
```

```python
print(f'x before any modification:\n {x}')
print()

print(f'x before any reshaping:\n {x_reshape}')
print()

x_reshape[0, 0] = 10
print(f'x_reshape after modifying its top left element:\n {x_reshape}')
print()

print(f'x after modifying x_reshape top left element:\n {x}')
```

## Code exercise

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 8a)
path = r'./College.csv'
college = pd.read_csv(path)
college.head()
```

```python
# 8b) 
college2 = pd.read_csv(path, index_col=0)
college3 = college.rename({'Unnamed: 0': 'College'}, axis=1)
college3 = college3.set_index('College')
```

```python
college2.head()
```

```python
college3.head()
```

```python
college = college3
```

```python
# 8c)
college.describe()
```

```python
# 8d
scatter_cols= ['Top10perc', 'Apps', 'Enroll']
pd.plotting.scatter_matrix(college[scatter_cols], alpha=0.2)
```

```python
# 8e)
ax = college.boxplot(column='Outstate', by='Private')
plt.suptitle('')
ax.set_title('Out-of-state tuition among public and state institutions')
plt.show()
```

```python
# 8f)
top_ten_perc = college['Top10perc']
threshold = [0, 50, 100]
labels = ['no', 'yes']
college['elite'] = pd.cut(top_ten_perc, bins=threshold, labels=labels)
```

```python
college['elite'].value_counts()
```

```python
ax = college.boxplot(column='Outstate', by='elite')
plt.suptitle('')
ax.set_title('Out-of-state tuition among perceived elite and non-elite institutions')
plt.show()
```

```python
# 8g)
college['Grad.Rate'] = college['Grad.Rate'].clip(upper=100)
ax = college.plot.hist(column=['Grad.Rate'], by="Private", figsize=(10, 8))
plt.suptitle('Graduation rate grouped by public vs private institutions')
plt.tight_layout()
```

```python
# 8h)
ax = college.boxplot(column='perc.alumni', by='Private')
plt.suptitle('')
ax.set_title('% of alumni who donate among public and state institutions')
plt.show()
```

```python
ax = college.boxplot(column='S.F.Ratio', by='Private')
plt.suptitle('')
ax.set_title('Student/faculty ratio among public and state institutions')
plt.show()
```
